```python
import pandas as pd
```

```python
df = pd.read_csv("/home/schaffen/Workspace/Project/lead-insight-AI/data/raw.csv")
df
```

```python
df = df.drop(["Location"],axis=1)
df
```

```python
df['LeadSource'].unique()
```

```python
df['ReferralSource'].unique()
```

```python
df['ReferralSource'] = df['ReferralSource'].replace('Twitter', 'Zalo')
df['ReferralSource'].unique()
```

```python
df.rename(columns={
    'Downloads': 'ProductViewCount'
}, inplace=True)
df.columns
```

```python
df.rename(columns={
    'FormSubmissions': 'InquiryCount'
}, inplace=True)
df.columns
```

```python
df.rename(columns={
    'ResponseTime (hours)': 'ResponseTime',
    'Conversion (Target)': 'Conversion',
    'TimeSpent (minutes)': 'TimeSpent',
    'FollowUpEmails':'FollowUpMessage'
}, inplace=True)
df.columns
```

```python
df['PaymentHistory'].unique()
```

```python
df['PaymentHistory'] = df['PaymentHistory'].replace('Good', '1')
df['PaymentHistory'] = df['PaymentHistory'].replace('No Payment', '0')
df['PaymentHistory'].unique()
```

```python
df['ResponseTime'] = df['ResponseTime']*60
```

```python
df['TimeSpent'] = df['TimeSpent']*60
```

```python
df = df.drop("EmailSent",axis=1)
```

```python
df.to_csv("processed.csv",index=False)
```
